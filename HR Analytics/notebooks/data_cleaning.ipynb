## Exploratory Data Analysis

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Attrition', data=df)
plt.title("Employee Attrition Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Salary vs Attrition
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df)
plt.title("Monthly Income vs Attrition")
plt.show()
```

Observation: Employees with lower salary show higher attrition.

Meaning: Salary is a key factor in employee retention.

Action: Company should review salary structure for low-income employees.

```python
#Overtime Impact
sns.countplot(x='OverTime', hue='Attrition', data=df)
plt.title("Overtime vs Attrition")
plt.show()
```

Observation: Employees working overtime are leaving more.

Meaning: High workload may be causing stress and dissatisfaction.

Action: Company should reduce overtime or improve work-life balance.

```python
sns.boxplot(x='Attrition', y='Age', data=df)
plt.title("Age vs Attrition")
plt.show()
```

Observation: Younger employees are leaving more frequently.

Meaning: They may lack career growth or long-term engagement.

Action: Provide better growth opportunities and training.

```python
#Department Wise Attrition
sns.countplot(x='Department', hue='Attrition', data=df)
plt.xticks(rotation=45)
plt.title("Department-wise Attrition")
plt.show()
```

Observation: Sales department has higher attrition.

Meaning: Certain departments have more workload or pressure.

Action: HR should investigate department-wise issues.

## Final Conclusion

The analysis shows that overtime, low salary, younger employees, and specific departments are the major factors affecting employee attrition. These insights can help HR teams improve employee retention strategies.
